### Reading the trace

Four things to look for:

* **Which agent is acting.** Five of them hand work to each other
* **Which tool it called**, and with what arguments
* **`changed state?`** marks the steps that changed something in the world
* **`reversible`** tells you whether it could be undone

The outcome block gives you four counts. **Actions that changed state** is what the agent actually did. Everything else was thinking.

> The **Run again (same day)** button re-runs without clearing the refund ledger, as if the customer came back. Some things only go wrong the second time.
