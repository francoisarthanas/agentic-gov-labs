## Step 1: Install


```python
#@title Step 1: Install { display-mode: "form" }
import io, os, shutil, subprocess, sys, urllib.request, zipfile

DIR  = "agentic-gov-labs"
REPO = "https://github.com/francoisarthanas/agentic-gov-labs.git"
ZIP  = ("https://github.com/francoisarthanas/agentic-gov-labs"
        "/archive/refs/heads/main.zip")


def _clone():
    r = subprocess.run(["git", "clone", "--depth", "1", REPO, DIR],
                       capture_output=True, text=True)
    if r.returncode != 0:
        print("git clone did not work here, downloading the zip instead.")
        with urllib.request.urlopen(ZIP, timeout=60) as resp:
            z = zipfile.ZipFile(io.BytesIO(resp.read()))
        z.extractall(".")
        shutil.move("agentic-gov-labs-main", DIR)


# A runtime you have used before may already hold an older copy of the labs.
# Bring it up to date, and start over if it will not come.
if os.path.isdir(DIR):
    fresh = False
    if os.path.isdir(os.path.join(DIR, ".git")):
        r = subprocess.run(["git", "-C", DIR, "pull", "--ff-only", "-q"],
                           capture_output=True, text=True)
        fresh = r.returncode == 0
    if not fresh or not os.path.isdir(os.path.join(DIR, "supportflow")):
        shutil.rmtree(DIR, ignore_errors=True)

if not os.path.isdir(DIR):
    try:
        _clone()
    except Exception as e:
        print("Could not download the labs:", e)
        print("Check your connection and run this cell again.")

if DIR not in sys.path:
    sys.path.insert(0, DIR)

for _m in [m for m in list(sys.modules) if m.split(".")[0] == "supportflow"]:
    del sys.modules[_m]                     # never import a half-old package

try:
    import supportflow as sf
    print(f"SupportFlow v{sf.__version__} ready. Move to Step 2.")
except ImportError as e:
    print("Not loaded yet:", e)
    print("Run this cell once more. It recovers on its own most times.")
```
