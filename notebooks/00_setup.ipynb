## Step 1: Install everything

**Runtime > Run all**, or run this cell on its own. About a minute.

If your network blocks `git`, this falls back to a direct download on its own. You do not need to do anything differently.


````python
#@title Step 1: Install { display-mode: "form" }
import io, os, shutil, subprocess, sys, urllib.request, zipfile

DIR  = "agentic-gov-labs"
REPO = "https://github.com/francoisarthanas/agentic-gov-labs.git"
ZIP  = ("https://github.com/francoisarthanas/agentic-gov-labs"
        "/archive/refs/heads/main.zip")

def _usable(d):
    return os.path.exists(os.path.join(d, "supportflow", "__init__.py"))

def _download():
    """Try git, then a plain zip download. Returns True if we got the code."""
    if _usable(DIR):
        print("SupportFlow is already here.")
        return True

    shutil.rmtree(DIR, ignore_errors=True)
    shutil.rmtree("agentic-gov-labs-main", ignore_errors=True)

    print("Downloading SupportFlow...")
    try:
        r = subprocess.run(["git", "clone", "--depth", "1", REPO, DIR],
                           capture_output=True, text=True, timeout=180)
        if r.returncode == 0 and _usable(DIR):
            return True
        detail = (r.stderr or r.stdout or "").strip().splitlines()
        print("  git could not download it here"
              + (": " + detail[-1] if detail else "."))
    except Exception as exc:
        print(f"  git is unavailable here ({type(exc).__name__}).")

    print("  Trying a direct download instead...")
    shutil.rmtree(DIR, ignore_errors=True)
    try:
        blob = urllib.request.urlopen(ZIP, timeout=180).read()
        zipfile.ZipFile(io.BytesIO(blob)).extractall(".")
        if os.path.exists("agentic-gov-labs-main"):
            os.rename("agentic-gov-labs-main", DIR)
        if _usable(DIR):
            print("  Direct download worked.")
            return True
    except Exception as exc:
        print(f"  Direct download failed too ({type(exc).__name__}).")
    return False

if not _download():
    print("\n" + "=" * 62)
    print("  COULD NOT DOWNLOAD SUPPORTFLOW")
    print("=" * 62)
    print("  This is a network problem, not something you did wrong.")
    print("  Try, in order:")
    print("    1. Runtime > Disconnect and delete runtime, then Run all")
    print("    2. Check you are not on a corporate VPN or proxy")
    print("    3. Post in Questions with a screenshot of this cell")
    print("=" * 62)
else:
    if DIR not in sys.path:
        sys.path.insert(0, DIR)
    print("Installing the console and the governance toolkit...")
    pip = subprocess.run(
        [sys.executable, "-m", "pip", "install", "-q",
         "-r", os.path.join(DIR, "requirements.txt")],
        capture_output=True, text=True)
    if pip.returncode != 0:
        tail = (pip.stderr or "").strip().splitlines()[-1:] or ["unknown"]
        print(f"  Some packages did not install ({tail[0][:70]}).")
        print("  Carry on anyway. Step 3 will tell you what is missing.")
    try:
        import supportflow as sf
        print(f"\nSupportFlow v{sf.__version__} ready. Move to Step 2.")
    except Exception as exc:
        print(f"\nDownloaded, but could not load it ({type(exc).__name__}).")
        print("Runtime > Restart session, then Run all.")
````
