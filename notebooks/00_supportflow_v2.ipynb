## Step 2: Connect your API key

You need this from Week 4. Set it up now, once, and Colab will remember it for every future lab.

**1.** Open [aistudio.google.com](https://aistudio.google.com) in a new tab and sign in with a **personal** Google account.

**2.** Click **Get API key**, then **Create API key**. Copy it.

**3. Read the terms before you accept them.** They differ between the free and paid tiers, and the difference is worth knowing. Bring what you find to the first session.

**4.** Back in Colab, click the **key icon** in the left sidebar to open **Secrets**.

**5.** Click **Add new secret**. Name it exactly:

```
GOOGLE_API_KEY
```

**6.** Paste your key as the value, and turn on **Notebook access**.

Then run the cell below to confirm Colab can read it.


````python
#@title Step 2: Check the key { display-mode: "form" }
import os

try:
    from google.colab import userdata
    try:
        key = userdata.get("GOOGLE_API_KEY")
        os.environ["GOOGLE_API_KEY"] = key
        print(f"Key found in Colab Secrets, ending {key[-4:]}.")
        print("Colab will remember it. You never have to paste it again.")
    except Exception:
        print("No secret named GOOGLE_API_KEY yet, or notebook access is off.")
        print("Open the key icon in the left sidebar and check both.")
        print("\nNothing before Week 4 needs it, so you can carry on.")
except ImportError:
    print("Not running in Colab. Set GOOGLE_API_KEY in your environment.")
````
