### Reading the trace

* **agent** — which of the five is acting
* **tool** — what it called
* **changed state?** — `YES` marks the steps that changed the world
* **reversible** — whether that change can be undone

The outcome block gives four counts. **Actions that changed state** is what the agent actually did. Everything else was thinking.
