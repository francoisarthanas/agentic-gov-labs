## Step 1: Set up

Run this cell once. It takes about twenty seconds.

```python
#@title Run this cell to set up { display-mode: "form" }
import os, sys, subprocess

REPO = "https://github.com/francoisarthanas/agentic-gov-labs.git"

if not os.path.exists("agentic-gov-labs"):
    print("Fetching SupportFlow...")
    subprocess.run(["git", "clone", "--quiet", "--depth", "1", REPO], check=True)

sys.path.insert(0, "agentic-gov-labs")

import supportflow as sf
print("SupportFlow v%s ready." % sf.__version__)
print("%d scenarios, %d tools, %d autonomy levels, %d controls."
      % (len(sf.scenarios.SCENARIOS), len(sf.tools.REGISTRY),
         len(sf.LEVELS), len(sf.Controls().as_dict())))
```

## Step 2: Open the Governance Console

This is the only interface you need.

| Control | What it does |
| :---- | :---- |
| **Scenario** | Which customer situation to run |
| **Autonomy** | How much the agent may do without a human. L1 to L5 |
| **Ceiling** | The hard refund cap enforced in the tool, in dollars |
| **Attack** | Arms one of the attack scenarios. Leave on `none` until Lab 4 |
| **Controls** | Eight independent safeguards. Turn them on and off one at a time |

Set what you want, then press **Run scenario**.

```python
#@title Governance Console { display-mode: "form" }
console = sf.console.launch()
```

### Reading the trace

Every line is one step. Lines marked `!!` **changed something in the world**.

Look for four things:

* Which agent acted. There are five of them
* Which tool it called, and with what arguments
* `[control]` lines, showing which safeguards evaluated and what they decided
* `[approval]` lines, showing where a human was required

At the bottom you get the outcome and four counts. **Actions that changed state** is the number that matters most.
